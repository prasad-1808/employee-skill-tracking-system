```python
import pandas as pd
import numpy as np
```

```python
# Load the data from the CSV files
employee_df = pd.read_csv(
    "C:/Users/Admin/Desktop/JMD327_MainProject/JMD327_JElite/PythonScripts/DataEngineering/RAW/raw_employee.csv"
)
skill_df = pd.read_csv(
    "C:/Users/Admin/Desktop/JMD327_MainProject/JMD327_JElite/PythonScripts/DataEngineering/RAW/raw_skill.csv"
)
course_df = pd.read_csv(
    "C:/Users/Admin/Desktop/JMD327_MainProject/JMD327_JElite/PythonScripts/DataEngineering/RAW/raw_course.csv"
)
```

```python
employee_table = employee_df[
    [
        "EmployeeID",
        "Firstname",
        "Lastname",
        "Designation",
        "YearOfJoining",
    ]
]

employee_table = employee_table.sort_values(by=['EmployeeID'])
employee_table.to_csv("prep_employee.csv", index=False)
```

```python

# Select the relevant columns
skill_table = skill_df[
    [
        "EmployeeID",
        "CourseID",
        "Proficiency",
        "SkillType",
        "CertificateLink",
        "ScoreObtained",
        "Verified",
    ]
]

# Sort the data by EmployeeID
skill_table = skill_table.sort_values(by=["EmployeeID"])

# Replace empty spaces (including those with just spaces) in CertificateLink with NaN
skill_table["CertificateLink"] = (
    skill_table["CertificateLink"].astype(str).str.strip().replace("", "NaN")
)

# Replace empty spaces and "nan" with actual NaN
skill_table["ScoreObtained"] = pd.to_numeric(
    skill_table["ScoreObtained"]
    .astype(str)
    .str.strip()
    .replace(["", "nan"], np.nan), 
    errors='coerce'
)

def categorize_score(score):
    if pd.isnull(score):
        return np.nan
    elif score > 90:
        return "Excellent"
    elif score > 80:
        return "Best"
    elif score > 70:
        return "Good"
    elif score > 60:
        return "Average"
    elif score > 50:
        return "Below average"
    else:
        return "Below average"


# Add ScoreCategory column using apply
skill_table["ScoreCategory"] = skill_table["ScoreObtained"].apply(categorize_score)


# Save the cleaned table to a CSV file
skill_table.to_csv("prep_skill.csv", index=False)
```

```python
course_table = course_df[
    [
        "CourseID",
        "CourseName",
        "Level",
        "CourseCode",
    ]
]

course_table = course_table.sort_values(by=["CourseID"])
course_table.to_csv("prep_course.csv", index=False)
```
